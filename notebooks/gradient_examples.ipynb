## 2. Visualizing Gradient Fields

A gradient field shows the gradient vector at every point in 2D space.

```python
# Visualize gradient field for f(x, y) = x² + y²
def f_2d(x, y):
    return x**2 + y**2

fig = visualize_gradient_field(f_2d, save_path='docs/images/gradient_field.png')
plt.show()
print("Gradient field visualization generated!")
```

## 3. Gradient Descent Optimization

Gradient descent finds the minimum of a function by moving in the negative gradient direction.

```python
# Run gradient descent example
print("=== Gradient Descent Example ===\n")
x_history, f_history = gradient_descent_example()

# Visualize gradient descent
def f(x):
    return x**2 + 2*x + 1
def df(x):
    return 2*x + 2

fig, _, _ = visualize_gradient_descent(f, df, x_start=5.0, learning_rate=0.1, 
                                      iterations=50, save_path='docs/images/gradient_descent.png')
plt.show()
```

## 4. Learning Rate Comparison

The learning rate controls step size. Too small = slow convergence, too large = overshoot.

```python
# Compare different learning rates
def f(x):
    return x**2 + 2*x + 1
def df(x):
    return 2*x + 2

fig = compare_learning_rates(f, df, x_start=5.0, 
                             learning_rates=[0.01, 0.1, 0.5, 1.0], 
                             iterations=30,
                             save_path='docs/images/learning_rates_comparison.png')
plt.show()
print("Learning rate comparison visualization generated!")
```

## 5. 2D Gradient Descent

Gradient descent works in multiple dimensions too!

```python
# 2D gradient descent: f(x, y) = x² + y² (minimum at (0, 0))
def f_2d(x, y):
    return x**2 + y**2

def df_dx(x, y):
    return 2*x

def df_dy(x, y):
    return 2*y

fig, x_history, y_history, f_history = gradient_descent_2d(
    f_2d, df_dx, df_dy, x_start=3.0, y_start=3.0, 
    learning_rate=0.1, iterations=30,
    save_path='docs/images/gradient_descent_2d.png'
)
plt.show()
print(f"Started at ({x_history[0]:.2f}, {y_history[0]:.2f})")
print(f"Ended at ({x_history[-1]:.4f}, {y_history[-1]:.4f})")
print(f"Final loss: {f_history[-1]:.6f}")
```
